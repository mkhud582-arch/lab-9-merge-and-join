## Normal Test Case

```python
import pandas as pd
import part2
```

```python
# Dataset Loading Test

main_df = pd.read_csv("Data/modcloth_main.csv")
prices_df = pd.read_csv("Data/modcloth_prices.csv")

print("Main Dataset Dimensions:", main_df.shape)
print("Prices Dataset Dimensions:", prices_df.shape)

print("\nMain Dataset:")
print(main_df.head())

print("\nModcloth Dataset:")
print(prices_df.head())

assert main_df.shape == (81790, 18)
assert prices_df.shape == (1378, 2)

print("\nDataset Loading Test: PASSED")
```

```python
# Matching Key Test

matching_key = main_df.columns.intersection(prices_df.columns)

print("Matching Key:", matching_key)

assert "item_id" in matching_key
assert list(matching_key) == ["item_id"]

print("Matching Key Test: PASSED")
```

```python
# Inner Merge Test

inner_merge = pd.merge(main_df, prices_df, on = "item_id", how = "inner")

assert inner_merge["item_id"].isin(main_df["item_id"]).all()
assert inner_merge["item_id"].isin(prices_df["item_id"]).all()

assert inner_merge.shape == (81790, 19)

print("Inner Merge Test: PASSED")
```

```python
# Left Merge Test

left_merge = pd.merge(main_df, prices_df, on = "item_id", how = "left")

assert left_merge["item_id"].isin(main_df["item_id"]).all()
assert len(left_merge) == len(main_df)
assert left_merge.shape == (81790, 19)

print("Left Merge Test: PASSED")
```

```python
# Right Merge Test

right_merge = pd.merge(main_df, prices_df, on = "item_id", how = "right")

assert right_merge["item_id"].isin(prices_df["item_id"]).all()
assert len(right_merge) >= len(prices_df)
assert right_merge.shape == (81798, 19)

print("Right Merge Test: PASSED")
```

```python
# Outer Merge Test 

outer_merge = pd.merge(main_df, prices_df, on = "item_id", how = "outer")

assert main_df["item_id"].isin(outer_merge["item_id"]).all()
assert prices_df["item_id"].isin(outer_merge["item_id"]).all()

assert outer_merge.shape == (81798, 19)

print("Outer Merge Test: PASSED")
```

```python
# Missing Values Test 

outer_merge = pd.merge(main_df, prices_df, on = "item_id", how = "outer")

# Identify records only in the main dataset
only_main_dataset = outer_merge[outer_merge["price"].isna()]

# Identify records only in the prices dataset
only_prices_dataset = outer_merge[outer_merge["size"].isna()]

# Verify unmatched record counts
assert len(only_main_dataset) == 0
assert len(only_prices_dataset) == 8

# Verify missing values are present in the unmatched records
assert only_prices_dataset["size"].isna().all()

print("Missing Values Test: PASSED")
```

## Special Test Case

```python
# Special Test Case: Duplicate Matching Keys

duplicate_keys = main_df[
    main_df["item_id"].duplicated()
]

print("Number of duplicate item_id records:",
      len(duplicate_keys))

assert len(duplicate_keys) > 0

print("Duplicate Matching Keys Test: PASSED")
```
